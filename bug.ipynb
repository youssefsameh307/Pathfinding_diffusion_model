```python
import torch 

conv = torch.nn.Conv2d(1, 1, 3, 1, 1)
inputs_sample = torch.randn(1, 1, 5, 5)
outputs_sample = conv(inputs_sample)
outputs_sample.device
```

```python
# Do with cuda
conv = conv.cuda()
inputs_sample = inputs_sample.cuda()
outputs_sample = conv(inputs_sample)
```
